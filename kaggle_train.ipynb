## 五组单变量消融

下面的 cell 会依次跑 5 组训练。**每组实时输出 stdout**（每步的 reward/KL/熵/hack rates）。

如果某组失败：**完整 stderr/traceback 会直接输出在 `=== [STDERR] preset: xxx ===` 下面**，然后自动跳到下一组继续，不会中断整个 notebook。

想快速验证就把 `MAX_STEPS` 改成 `5`。

```python
# ============================================================
# Cell 6: 七组单变量消融（唯一需要改的配置区）
# 防挂死设计（上一版 12h 白烧的教训）：
#   1. --use-swanlab false  ← 根因修复：swanlab 联网认证会永久挂起
#   2. stdin=DEVNULL + stderr 后台线程实时抽干 → 管道永不阻塞
#   3. 看门狗：10 分钟无输出 → 杀进程、打印 stderr、转下一组
#   4. 每组预算 95 分钟 → 超时杀进程、保留已训练曲线、转下一组
#   5. 全局截止 10.5h → 跳过剩余组，保证 Cell 7 一定出图
# ============================================================
PRESETS = ["clip_higher"]   # 单组补跑：run-2 该组 OOM@step70，其余 4 组数据沿用 run-2
MAX_STEPS = 150         # 她的叙事：坍缩发生在 150 步内
DATASET = "gsm8k"       # gsm8k | gsm8k_math
TRAIN_SAMPLES = 500
EVAL_SAMPLES = 200      # 与 run-2 其余 4 组严格同口径（可比性优先）
PPO_EPOCHS = 4          # 关键：让 clip 真正生效（r 偏离 1），激活熵压制->clip_higher 救熵机制链
LEARNING_RATE = "2e-5"  # 配合 4 epoch 把策略推进不稳定区，触发基线退化
OVERLONG_CACHE = 64     # 关键修正：惩罚窗口 128->64（区[192,256]），对齐论文 25% 比例，
                        # 消除"空输出=局部最优"的死亡螺旋
QUIET_KILL_MIN = 10     # 连续多少分钟无输出判定挂死
BUDGET_MIN = 130        # 救熵后输出变长变慢，单组放宽到 130min
GLOBAL_DEADLINE_H = 10.5
# ============================================================

import subprocess, time, json, sys, os, traceback
import threading, queue

REPO = "/kaggle/working/qwen-math-grpo-dapo"
results = {}
env = os.environ.copy()
env["PYTHONUNBUFFERED"] = "1"
env["HF_HUB_DOWNLOAD_TIMEOUT"] = "30"
env["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"  # OOM 修复：消除显存碎片累积
T_GLOBAL0 = time.time()

def save_results():
    with open("/kaggle/working/ablation_results.json", "w") as f:
        json.dump(results, f, indent=2)

def parse_output(lines):
    curve, final, eval_vals = [], {}, {}
    surprisal_curve, zerovar_curve, len_curve = [], [], []
    for line in lines:
        line = line.strip()
        if line.startswith("[step"):
            body = line.split("] ", 1)[-1] if "] " in line else ""
            vals = {}
            for part in body.split(" | "):
                k, _, v = part.partition("=")
                try:
                    vals[k] = float(v)
                except ValueError:
                    pass
            final = vals
            if "mean" in vals:
                curve.append(vals["mean"])
            if "mean_sampled_token_surprisal" in vals:
                surprisal_curve.append(vals["mean_sampled_token_surprisal"])
            if "zero_variance_group_ratio" in vals:
                zerovar_curve.append(vals["zero_variance_group_ratio"])
            if "mean_completion_tokens" in vals:
                len_curve.append(vals["mean_completion_tokens"])
        elif line.startswith("[eval]"):
            for part in line.split("] ", 1)[-1].split(" "):
                k, _, v = part.partition("=")
                if k in ("accuracy", "format_rate") or k.startswith("pass@"):
                    try:
                        eval_vals[k] = float(v)
                    except ValueError:
                        pass
    final["_surprisal_curve"] = surprisal_curve
    final["_zerovar_curve"] = zerovar_curve
    final["_len_curve"] = len_curve
    return curve, final, eval_vals

def record(algo, t0, lines, status, err_tail):
    curve, final, eval_vals = parse_output(lines)
    results[algo] = {
        "minutes": round((time.time() - t0) / 60, 1),
        "status": status,
        "accuracy": eval_vals.get("accuracy"),
        "format_rate": eval_vals.get("format_rate"),
        "pass@8": eval_vals.get("pass@8"),
        "final_reward": final.get("mean"),
        "surprisal": final.get("mean_sampled_token_surprisal"),
        "surprisal_curve": final.get("_surprisal_curve", []),
        "zero_var_curve": final.get("_zerovar_curve", []),
        "len_curve": final.get("_len_curve", []),
        "clip_fraction": final.get("clip_fraction"),
        "zero_var_ratio": final.get("zero_variance_group_ratio"),
        "mean_completion_tokens": final.get("mean_completion_tokens"),
        "reward_curve": curve,
        "stderr_tail": err_tail,
    }
    save_results()
    print(f"\n[summary] {algo}: status={status} acc={results[algo]['accuracy']} "
          f"pass@8={results[algo]['pass@8']} steps={len(curve)} "
          f"({results[algo]['minutes']} min)", flush=True)

for algo in PRESETS:
    if time.time() - T_GLOBAL0 > GLOBAL_DEADLINE_H * 3600:
        print(f">>> [skip] {algo}: 已过全局截止，把剩余额度留给出图", flush=True)
        results[algo] = {"status": "skipped_global_deadline", "reward_curve": []}
        save_results()
        continue

    print(f"\n{'='*70}\n>>> preset: {algo}\n{'='*70}", flush=True)
    cmd = [sys.executable, "scripts/train.py",
           "--algorithm", algo, "--dataset", DATASET,
           "--train-samples", str(TRAIN_SAMPLES),
           "--eval-samples", str(EVAL_SAMPLES),
           "--pass-at-k", "8", "--pass-k-samples", "50",
           "--max-steps", str(MAX_STEPS),
           "--ppo-epochs", str(PPO_EPOCHS),
           "--learning-rate", LEARNING_RATE,
           "--overlong-cache", str(OVERLONG_CACHE),
           "--device", "cuda",
           "--use-swanlab", "false",
           "--run-name", algo]
    t0 = time.time()
    try:
        proc = subprocess.Popen(cmd, cwd=REPO,
                                stdout=subprocess.PIPE,
                                stderr=subprocess.PIPE,
                                stdin=subprocess.DEVNULL,
                                text=True, bufsize=1, env=env)
        out_q, err_lines = queue.Queue(), []

        def pump_out():
            for line in proc.stdout:
                out_q.put(line.rstrip("\n"))
            out_q.put(None)  # EOF 标记

        def pump_err():
            for line in proc.stderr:
                err_lines.append(line.rstrip("\n"))

        threading.Thread(target=pump_out, daemon=True).start()
        threading.Thread(target=pump_err, daemon=True).start()

        stdout_lines, last_active, kill_reason = [], time.time(), None
        while True:
            try:
                line = out_q.get(timeout=15)
            except queue.Empty:
                line = "__timeout__"
            now = time.time()
            if line is None:
                break  # 子进程关闭了 stdout，正常结束路径
            if line != "__timeout__":
                print(line, flush=True)
                stdout_lines.append(line)
                last_active = now
                continue
            if now - last_active > QUIET_KILL_MIN * 60:
                kill_reason = f"watchdog: {QUIET_KILL_MIN} 分钟无输出，判定挂死"
                break
            if now - t0 > BUDGET_MIN * 60:
                kill_reason = f"budget: 超过 {BUDGET_MIN} 分钟（保留已训练曲线）"
                break

        if kill_reason:
            print(f"\n=== [KILL] {algo}: {kill_reason} ===", flush=True)
            proc.kill()
        try:
            proc.wait(timeout=60)
        except subprocess.TimeoutExpired:
            proc.kill()
            proc.wait()
        time.sleep(1.0)  # 等 stderr 泵收尾

        if proc.returncode != 0 or kill_reason:
            print(f"=== [STDERR tail] {algo} 最后 40 行 ===", flush=True)
            print("\n".join(err_lines[-40:]) or "(stderr 为空)", flush=True)
            if kill_reason and kill_reason.startswith("budget"):
                status = f"partial: {kill_reason}"
            else:
                status = f"dead: {kill_reason or f'exit code {proc.returncode}'}"
            record(algo, t0, stdout_lines, status, "\n".join(err_lines[-20:]))
            continue

        if err_lines:
            print(f"=== [STDERR tail] {algo} 最后 5 行（正常完成，仅警告） ===", flush=True)
            print("\n".join(err_lines[-5:]), flush=True)
        record(algo, t0, stdout_lines, "ok", "\n".join(err_lines[-5:]))

    except Exception as e:
        print(f"\n=== [FATAL] preset: {algo} ===", flush=True)
        traceback.print_exc()
        results[algo] = {"status": f"fatal: {e}", "reward_curve": [],
                         "minutes": round((time.time() - t0) / 60, 1)}
        save_results()
        continue

print(f"\n{'='*70}")
print(f"全部结束，总耗时 {(time.time() - T_GLOBAL0) / 3600:.2f} h")
print("结果 -> /kaggle/working/ablation_results.json")
print(f"{'='*70}", flush=True)
```

```python
# Cell 7: 画图 + 结果汇总
import json
import pandas as pd
import matplotlib.pyplot as plt

with open("/kaggle/working/ablation_results.json") as f:
    results = json.load(f)

print("=== 完整结果表 ===")
df = pd.DataFrame(results).T
display(df[["accuracy", "pass@8", "format_rate", "final_reward", "surprisal",
            "clip_fraction", "zero_var_ratio", "mean_completion_tokens", "minutes"]])

success = {a: r for a, r in results.items() if r.get("accuracy") is not None}
failed = {a: r for a, r in results.items() if r.get("accuracy") is None}
print(f"\n成功: {len(success)} 组, 失败: {len(failed)} 组")
if failed:
    for a, r in failed.items():
        print(f"  FAIL {a}: {r.get('error', 'unknown')} ({r.get('minutes', 0)} min)")
        if r.get("stderr_tail"):
            print(f"    stderr tail: {r['stderr_tail'][-200:]}")

fig, axes = plt.subplots(1, 3, figsize=(20, 5))

for algo, r in results.items():
    if r.get("reward_curve"):
        axes[0].plot(r["reward_curve"], label=algo, alpha=0.85)
axes[0].set_xlabel("step")
axes[0].set_ylabel("mean reward")
axes[0].set_title("Training reward curves")
if success:
    axes[0].legend(fontsize=8)
axes[0].grid(alpha=0.3)

algos = list(results.keys())
accs = [results[a]["accuracy"] or 0 for a in algos]
colors = ["#9e9e9e"] * len(algos)
if "dapo" in algos:
    colors[algos.index("dapo")] = "#d62728"
if "entropy_reg" in algos:
    colors[algos.index("entropy_reg")] = "#1f77b4"
axes[1].bar(algos, accs, color=colors)
axes[1].set_ylabel("GSM8K accuracy")
axes[1].set_title("Final eval accuracy")
axes[1].tick_params(axis="x", rotation=30)
axes[1].grid(alpha=0.3, axis="y")

# 熵代理曲线：策略 surprisal 随训练步数的演变（复刻叙事的核心图）
for algo, r in results.items():
    sc = r.get("surprisal_curve") or []
    if sc:
        axes[2].plot(sc, label=algo, alpha=0.85)
axes[2].set_xlabel("step")
axes[2].set_ylabel("mean sampled token surprisal")
axes[2].set_title("Policy surprisal (entropy proxy) over training")
if success:
    axes[2].legend(fontsize=8)
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("/kaggle/working/ablation_summary.png", dpi=150)
print("saved -> /kaggle/working/ablation_summary.png")
plt.show()
```
